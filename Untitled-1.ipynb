

Denoising Autoencoder for Noise Removal + Performance Comparison
- Noise: Gaussian, Salt & Pepper
- Latent sizes: 16 and 64
- Metrics: MSE, PSNR
- Visuals: side-by-side input/output
"""

```python
import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow import keras
from tensorflow.keras import layers
```


 Repro & GPU prints


```python
np.random.seed(42)
tf.random.set_seed(42)
print("TensorFlow:", tf.__version__)
print("GPU Available:", tf.config.list_physical_devices('GPU'))
```

 Data: Fashion-MNIST

```python
(x_train, _), (x_test, _) = keras.datasets.fashion_mnist.load_data()
x_train = (x_train.astype("float32") / 255.0)[..., np.newaxis]  # (N, 28, 28, 1)
x_test  = (x_test.astype("float32")  / 255.0)[..., np.newaxis]
print("Train:", x_train.shape, "Test:", x_test.shape)
```

2) Noise functions

```python
def add_gaussian_noise(x, sigma=0.2):
    """x in [0,1], sigma ~ 0..1"""
    noise = tf.random.normal(shape=tf.shape(x), mean=0.0, stddev=sigma, dtype=x.dtype)
    y = tf.clip_by_value(x + noise, 0.0, 1.0)
    return y

def add_salt_pepper_noise(x, amount=0.2, salt_vs_pepper=0.5):
    """
    amount: fraction of pixels to corrupt
    salt_vs_pepper: fraction of salt among corrupted (rest are pepper)
    """
    x = tf.convert_to_tensor(x)
    h = tf.shape(x)[1]
    w = tf.shape(x)[2]
    c = tf.shape(x)[3]

    num_pixels = tf.cast(h * w * c, tf.float32)
    num_corrupt = tf.cast(tf.round(amount * num_pixels), tf.int32)
    num_salt = tf.cast(tf.round(salt_vs_pepper * tf.cast(num_corrupt, tf.float32)), tf.int32)
    num_pepper = num_corrupt - num_salt

    # Flatten per image, create salt/pepper indices
    x_flat = tf.reshape(x, (tf.shape(x)[0], -1))
    n = tf.shape(x_flat)[0]

    def corrupt_one(img):
        idx = tf.range(tf.shape(img)[0])
        idx = tf.random.shuffle(idx)
        salt_idx = idx[:num_salt]
        pepper_idx = idx[num_salt:num_salt+num_pepper]

        img = tf.tensor_scatter_nd_update(img, tf.expand_dims(salt_idx, 1), tf.ones([num_salt], dtype=img.dtype))
        img = tf.tensor_scatter_nd_update(img, tf.expand_dims(pepper_idx, 1), tf.zeros([num_pepper], dtype=img.dtype))
        return img

    x_cor = tf.map_fn(corrupt_one, x_flat)
    x_cor = tf.reshape(x_cor, tf.shape(x))
    return x_cor
```

3) Model builder (Conv AE with Dense latent)

```python
def build_dae(latent_dim=64, input_shape=(28, 28, 1)):
    inp = layers.Input(shape=input_shape)

    # Encoder: conv downsampling to 7x7, then Dense(latent_dim)
    x = layers.Conv2D(32, 3, strides=2, padding="same", activation="relu")(inp)   # 14x14
    x = layers.Conv2D(16, 3, strides=2, padding="same", activation="relu")(x)     # 7x7
    shape_before_flatten = tf.keras.backend.int_shape(x)[1:]  # (7, 7, 16)
    x = layers.Flatten()(x)
    latent = layers.Dense(latent_dim, activation="relu", name="latent")(x)

    # Decoder: Dense -> reshape -> transposed conv back to 28x28x1
    x = layers.Dense(np.prod(shape_before_flatten), activation="relu")(latent)
    x = layers.Reshape(shape_before_flatten)(x)
    x = layers.Conv2DTranspose(16, 3, strides=2, padding="same", activation="relu")(x)  # 14x14
    x = layers.Conv2DTranspose(32, 3, strides=2, padding="same", activation="relu")(x)  # 28x28
    out = layers.Conv2D(1, 3, padding="same", activation="sigmoid")(x)

    model = keras.Model(inp, out, name=f"dae_latent{latent_dim}")
    model.compile(optimizer="adam", loss="mse")
    return model
```

4) Training utility

```python
def train_on_noise(noise_type, latent_dim, epochs=5, batch_size=128,
                   sigma=0.2, amount=0.2, salt_vs_pepper=0.5):
    """
    noise_type: 'gaussian' or 'sp'
    Returns: trained_model, (x_test_noisy), history
    """
    # Make noisy inputs
    if noise_type == "gaussian":
        x_train_noisy = add_gaussian_noise(x_train, sigma=sigma)
        x_test_noisy  = add_gaussian_noise(x_test,  sigma=sigma)
    elif noise_type == "sp":
        x_train_noisy = add_salt_pepper_noise(x_train, amount=amount, salt_vs_pepper=salt_vs_pepper)
        x_test_noisy  = add_salt_pepper_noise(x_test,  amount=amount, salt_vs_pepper=salt_vs_pepper)
    else:
        raise ValueError("noise_type must be 'gaussian' or 'sp'")

    model = build_dae(latent_dim=latent_dim, input_shape=(28,28,1))
    print("\n==== Training:", model.name, "Noise:", noise_type, "====")
    hist = model.fit(x_train_noisy, x_train,
                     validation_data=(x_test_noisy, x_test),
                     epochs=epochs, batch_size=batch_size, shuffle=True, verbose=2)
    return model, x_test_noisy, hist
```

5) Metrics: MSE & PSNR

```python
def mse_psnr(clean, denoised):
    # clean, denoised in [0,1]
    clean_tf = tf.convert_to_tensor(clean)
    den_tf = tf.convert_to_tensor(denoised)
    mse = tf.reduce_mean(tf.square(clean_tf - den_tf)).numpy().item()
    # tf.image.psnr computes per-image PSNR; take mean
    psnr_vals = tf.image.psnr(den_tf, clean_tf, max_val=1.0)
    psnr = tf.reduce_mean(psnr_vals).numpy().item()
    return mse, psnr
```

# 6) Run experiments

```python
EPOCHS = 15        # increase to 10–20 for stronger results
BATCH  = 128
SIGMA  = 0.2      # Gaussian std
AMOUNT = 0.2      # S&P corruption fraction
SVP    = 0.5      # fraction of salt among corruptions

results = []  # (noise, latent, test_mse, test_psnr)

# a) Gaussian: latent 16 vs 64
g_model_16, g_xnoisy, _ = train_on_noise("gaussian", 16, EPOCHS, BATCH, sigma=SIGMA)
g_pred_16 = g_model_16.predict(g_xnoisy, verbose=0)
g_mse16, g_psnr16 = mse_psnr(x_test, g_pred_16)
results.append(("Gaussian", 16, g_mse16, g_psnr16))

g_model_64, g_xnoisy2, _ = train_on_noise("gaussian", 64, EPOCHS, BATCH, sigma=SIGMA)
g_pred_64 = g_model_64.predict(g_xnoisy2, verbose=0)
g_mse64, g_psnr64 = mse_psnr(x_test, g_pred_64)
results.append(("Gaussian", 64, g_mse64, g_psnr64))

# b) Salt & Pepper: latent 16 vs 64
sp_model_16, sp_xnoisy, _ = train_on_noise("sp", 16, EPOCHS, BATCH, amount=AMOUNT, salt_vs_pepper=SVP)
sp_pred_16 = sp_model_16.predict(sp_xnoisy, verbose=0)
sp_mse16, sp_psnr16 = mse_psnr(x_test, sp_pred_16)
results.append(("Salt&Pepper", 16, sp_mse16, sp_psnr16))

sp_model_64, sp_xnoisy2, _ = train_on_noise("sp", 64, EPOCHS, BATCH, amount=AMOUNT, salt_vs_pepper=SVP)
sp_pred_64 = sp_model_64.predict(sp_xnoisy2, verbose=0)
sp_mse64, sp_psnr64 = mse_psnr(x_test, sp_pred_64)
results.append(("Salt&Pepper", 64, sp_mse64, sp_psnr64))
```

7) Print comparison table

```python
print("\n================ Performance Comparison (Test) ================")
print("{:<13} {:>8} {:>12} {:>10}".format("Noise", "Latent", "MSE", "PSNR(dB)"))
for noise, ld, mse, psnr in results:
    print("{:<13} {:>8} {:>12.6f} {:>10.3f}".format(noise, ld, mse, psnr))
print("===============================================================\n")
```

8) Visualizations

```python
def show_grid(title, originals, noisy, den_16, den_64, n=10, save_as=None):
    plt.figure(figsize=(20, 6))
    for i in range(n):
        # Row 1: original
        ax = plt.subplot(4, n, i + 1)
        plt.imshow(np.squeeze(originals[i]), cmap="gray")
        ax.set_title("Original")
        ax.axis("off")

        # Row 2: noisy
        ax = plt.subplot(4, n, i + 1 + n)
        plt.imshow(np.squeeze(noisy[i]), cmap="gray")
        ax.set_title("Noisy")
        ax.axis("off")

        # Row 3: denoised (latent 16)
        ax = plt.subplot(4, n, i + 1 + 2*n)
        plt.imshow(np.squeeze(den_16[i]), cmap="gray")
        ax.set_title("Denoised L16")
        ax.axis("off")

        # Row 4: denoised (latent 64)
        ax = plt.subplot(4, n, i + 1 + 3*n)
        plt.imshow(np.squeeze(den_64[i]), cmap="gray")
        ax.set_title("Denoised L64")
        ax.axis("off")

    plt.suptitle(title, y=0.99, fontsize=16)
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    if save_as:
        plt.savefig(save_as, dpi=150)
    plt.show()

# Gaussian grid
show_grid(
    title=f"Gaussian Noise σ={SIGMA} — Comparison (Latent 16 vs 64)",
    originals=x_test,
    noisy=g_xnoisy,
    den_16=g_pred_16,
    den_64=g_pred_64,
    n=10,
    save_as="grid_gaussian_compare.png"
)

# Salt & Pepper grid
show_grid(
    title=f"Salt&Pepper Noise amount={AMOUNT}, SvP={SVP} — Comparison (Latent 16 vs 64)",
    originals=x_test,
    noisy=sp_xnoisy,
    den_16=sp_pred_16,
    den_64=sp_pred_64,
    n=10,
    save_as="grid_sp_compare.png"
)
```
